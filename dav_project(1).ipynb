**Replacing missing values**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("student_academic_placement_performance_dataset(1).csv")

# Check missing values before replacement
print("Missing values before:")
print(df.isnull().sum())

# Replace missing values using recommended direct reassignment to avoid FutureWarnings
for column in df.columns:
    if df[column].dtype == 'object':
        # For text/categorical columns → replace with mode and assign back
        df[column] = df[column].fillna(df[column].mode()[0])
    else:
        # For numerical columns → replace with median and assign back
        df[column] = df[column].fillna(df[column].median())

# Check missing values after replacement
print("\nMissing values after:")
print(df.isnull().sum())
```

**Check invalid percentage values**

```python
percentage_columns = [
    "ssc_percentage",
    "hsc_percentage",
    "degree_percentage",
    "entrance_exam_score",
    "technical_skill_score",
    "soft_skill_score",
    "attendance_percentage"
]

print("\nChecking Percentage Columns:")

for column in percentage_columns:
    invalid = df[(df[column] < 0) | (df[column] > 100)]

    print(column, "-> Invalid values:", len(invalid))
```

**Check CGPA**

**Check negative values**

```python
numeric_columns = [
    "internship_count",
    "live_projects",
    "work_experience_months",
    "certifications",
    "backlogs",
    "salary_package_lpa"
]

print("\nChecking Negative Values:")

for column in numeric_columns:
    negative = (df[column] < 0).sum()
    print(column, "-> Negative values:", negative)
```

**Check placement status**

```python
invalid_placement = df[
    ~df["placement_status"].isin([0, 1])
]

print("\nInvalid Placement Status:", len(invalid_placement))

if len(invalid_placement) > 0:
    print(invalid_placement)
```

**Check logical consistency of salary**

```python
# Placed students with zero salary
placed_zero_salary = df[
    (df["placement_status"] == 1) &
    (df["salary_package_lpa"] <= 0)
]

# Unplaced students with salary
unplaced_salary = df[
    (df["placement_status"] == 0) &
    (df["salary_package_lpa"] > 0)
]

print("\nPlaced students with zero salary:",
      len(placed_zero_salary))

print("Unplaced students with salary:",
      len(unplaced_salary))
```

**Check outliers using IQR**

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("\nOutlier Analysis:")

for column in numeric_columns:

    # Don't treat Student ID as an analytical variable
    if column == "student_id":
        continue

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(column, "->", len(outliers), "outliers")
```

**Remove duplicates**

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("\nDuplicate rows removed:", before - after)
print("Rows after removing duplicates:", after)
```

**Handle missing values**

```python
numeric_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include="object").columns

# Fill numerical missing values with median
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

# Fill categorical missing values with mode
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

**Final data validation**

```python
print("\n========== FINAL DATA VALIDATION ==========")

print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))

print("Total Missing Values:",
      df.isnull().sum().sum())

print("Total Duplicate Rows:",
      df.duplicated().sum())

print("Duplicate Student IDs:",
      df["student_id"].duplicated().sum())
```

**Save the cleaned dataset**

```python
df.to_csv("cleaned_student_dataset.csv", index=False)

print("\nCleaned dataset saved successfully!")
```

**Generate a complete data-quality report**

```python
report = []

for column in df.columns:

    missing = df[column].isnull().sum()
    unique = df[column].nunique()

    report.append({
        "Column": column,
        "Data Type": str(df[column].dtype),
        "Missing Values": missing,
        "Missing %": round((missing / len(df)) * 100, 2),
        "Unique Values": unique
    })

quality_report = pd.DataFrame(report)

print("\n========== DATA QUALITY REPORT ==========")
print(quality_report)
```
