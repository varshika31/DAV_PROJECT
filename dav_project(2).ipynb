```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_student_dataset.csv")

gender_counts = df["gender"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["#4C78A8", "#F58518"],
    wedgeprops={"width": 0.42, "edgecolor": "white", "linewidth": 2},
    textprops={"fontsize": 12}
)

plt.text(
    0, 0,
    f"{len(df):,}\nStudents",
    ha="center",
    va="center",
    fontsize=18,
    fontweight="bold"
)

plt.title(
    "Gender Distribution of Students",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_student_dataset.csv")

bins = [0, 20, 40, 60, 80, 100]
labels = ["0–20", "21–40", "41–60", "61–80", "81–100"]

df["technical_skill_range"] = pd.cut(
    df["technical_skill_score"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

counts = df["technical_skill_range"].value_counts().reindex(labels)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    labels,
    counts.values,
    color=["#FFB000", "#FF7C43", "#F95D6A", "#A05195", "#665191"],
    edgecolor="black",
    linewidth=0.8
)

for bar, value in zip(bars, counts.values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        value + 10,
        str(value),
        ha="center",
        fontweight="bold"
    )

plt.title("Students by Technical Skill Score Range",
          fontsize=18, fontweight="bold")

plt.xlabel("Technical Skill Score Range")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_student_dataset.csv")

soft_bins = [0, 20, 40, 60, 80, 100]
soft_labels = ["0–20", "21–40", "41–60", "61–80", "81–100"]

df["soft_skill_range"] = pd.cut(
    df["soft_skill_score"],
    bins=soft_bins,
    labels=soft_labels,
    include_lowest=True
)

counts = df["soft_skill_range"].value_counts().reindex(soft_labels)

x = range(len(soft_labels))

plt.figure(figsize=(10, 6))

plt.fill_between(
    x,
    counts.values,
    alpha=0.45,
    color="#00A6A6"
)

plt.plot(
    x,
    counts.values,
    marker="o",
    linewidth=3,
    color="#005F73"
)

plt.xticks(x, soft_labels)

plt.title("Students Across Soft Skill Score Ranges",
          fontsize=18, fontweight="bold")

plt.xlabel("Soft Skill Score Range")
plt.ylabel("Number of Students")

plt.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_student_dataset.csv")

bins = [0, 5, 6, 7, 8, 9, 10]
labels = ["<5", "5–6", "6–7", "7–8", "8–9", "9–10"]

df["CGPA_Range"] = pd.cut(
    df["cgpa"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

placement_table = pd.crosstab(
    df["CGPA_Range"],
    df["placement_status"]
).reindex(labels)

placement_table.columns = ["Not Placed", "Placed"]

plt.figure(figsize=(11, 6))

placement_table.plot(
    kind="bar",
    stacked=True,
    figsize=(11, 6),
    color=["#FF595E", "#1982C4"],
    edgecolor="white"
)

plt.title("Placement Status Across CGPA Ranges",
          fontsize=18, fontweight="bold")

plt.xlabel("CGPA Range")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)
plt.legend(title="Placement Status")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_student_dataset.csv")

experience_table = pd.crosstab(
    df["work_experience_months"],
    df["placement_status"]
).sort_index()

experience_table.columns = ["Not Placed", "Placed"]

plt.figure(figsize=(11, 6))

plt.plot(
    experience_table.index,
    experience_table["Placed"],
    marker="o",
    linewidth=2.5,
    color="#06D6A0",
    label="Placed"
)

plt.plot(
    experience_table.index,
    experience_table["Not Placed"],
    marker="s",
    linewidth=2.5,
    color="#EF476F",
    label="Not Placed"
)

plt.title("Placement Distribution Across Work Experience",
          fontsize=18, fontweight="bold")

plt.xlabel("Work Experience (Months)")
plt.ylabel("Number of Students")

plt.legend()
plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```
